```python
from kanren import *
from kanren import lall, run, membero,conde, vars
from kanren.constraints import neq
```

```python
"""
knight-> always tells the truth
knave-> always lies
John says: "David is a knave."
David says: "I know that John is a knight and that Michael is a knave."
Michael says: "I and David are different."
"""
```

```python
john, david, michael = vars(3)
```

```python
kb = lall(
    membero(john,('knight','knave')),
    membero(david,('knight','knave')),
    membero(michael,('knight','knave')),
    #John says: "David is a knave.": if John is knight, then David is knave: david=knave, but if John is knave, then david is a knignt
    conde(
        [eq(john,'knight'),eq(david,'knave')],
        [eq(john,'knave'),eq(david,'knight')]
    ),
    #David says: "I know that John is a knight and that Michael is a knave.":
    #if David is knight, then john is knight and michael is knave
    #if david is knave,then john is knave and michael is knight
    conde(
        [eq(david, 'knight'), eq(john, 'knight'), eq(michael, 'knave')],
        [eq(david, 'knave'),  eq(john, 'knave')],  
        [eq(david, 'knave'),  eq(michael, 'knight')] 
    ),
    #Michael says: "I and David are different.":
    #if michael is knight, then david!=michael
    #if michael is knave,then david=michael(they are same)
    conde(
        [eq(michael,'knight'),neq(david,michael)],
        [eq(michael,'knave'),eq(david,michael)]
    )
)
```

```python
solutions = run(0, (john, david, michael), kb)
john, david, michael = solutions[0]
print('John is a', john)
print('David is a', david)
print('Michael is a', michael)
```
